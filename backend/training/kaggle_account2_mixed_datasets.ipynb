```python
# ── CELL 2: Imports + config ─────────────────────────────────
import os, json, re, time, warnings
from datetime import datetime
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from datasets import load_dataset, Dataset
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import (accuracy_score, f1_score, precision_score,
                              recall_score, classification_report)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.calibration import CalibratedClassifierCV
from transformers import (
    AutoTokenizer, AutoModelForSequenceClassification,
    TrainingArguments, Trainer, DataCollatorWithPadding,
    EarlyStoppingCallback, get_cosine_schedule_with_warmup,
    pipeline as hf_pipeline,
)
from datasketch import MinHash, MinHashLSH
warnings.filterwarnings('ignore')

assert torch.cuda.is_available(), 'ERROR: No GPU! Enable T4 in Settings → Accelerator'
print(f' GPU: {torch.cuda.get_device_name(0)}')
print(f'   VRAM: {torch.cuda.get_device_properties(0).total_memory/1e9:.1f} GB')

# HF Token — add via Add-ons → Secrets → HF_TOKEN
HF_TOKEN = ''
try:
    from kaggle_secrets import UserSecretsClient
    HF_TOKEN = UserSecretsClient().get_secret('HF_TOKEN')
    os.environ['HF_TOKEN'] = HF_TOKEN
    print(' HF token loaded')
except Exception:
    HF_TOKEN = os.environ.get('HF_TOKEN', '')
    print(' HF token from env' if HF_TOKEN else 'WARNING:  No HF token')

if HF_TOKEN:
    try:
        from huggingface_hub import login
        login(token=HF_TOKEN, add_to_git_credential=False)
        print(' HuggingFace login OK')
    except Exception as e:
        print(f'WARNING:  HF login failed: {e}')

OUTPUT_DIR = '/kaggle/working/model_B'
CKPT_DIR   = '/kaggle/working/checkpoints_B'
MODEL_NAME = 'roberta-base'
BATCH_SIZE = 32
SEED       = 42

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(CKPT_DIR,   exist_ok=True)
torch.manual_seed(SEED)
np.random.seed(SEED)
print(' Config ready')
```

```python
# ── CELL 3: Load all datasets ────────────────────────────────
# All IDs verified from actual Kaggle run outputs.
# No gated datasets. No custom scripts except with trust_remote_code=True.

frames = []

def add_frame(name, df):
    df = df.copy()
    df['source'] = name
    df = df[['text', 'label', 'source']]
    frames.append(df)
    print(f'    {name}: {len(df):,}')

def norm_label(series):
    raw = series.astype(str).str.lower().str.strip()
    mapped = raw.map({'1':1,'0':0,'fake':1,'real':0,'true':0,'false':1,
                       'yes':1,'no':0,'1.0':1,'0.0':0})
    return pd.to_numeric(mapped, errors='coerce')

# ── 1. GonzaloA (24k) — confirmed working ──
print('1.  GonzaloA/fake_news...')
try:
    ds = load_dataset('GonzaloA/fake_news', split='train').to_pandas()
    if 'title' in ds.columns:
        ds['text'] = (ds['title'].fillna('') + ' ' + ds['text'].fillna('')).str.strip()
    ds['label'] = ds['label'].astype(int)
    add_frame('GonzaloA', ds)
except Exception as e:
    print(f'   ERROR: {e}')

# ── 2. WELFake (72k) — confirmed working ──
print('2.  davanstrien/WELFake...')
try:
    ds = load_dataset('davanstrien/WELFake', split='train').to_pandas()
    if 'title' in ds.columns and 'text' in ds.columns:
        ds['text'] = (ds['title'].fillna('') + ' ' + ds['text'].fillna('')).str.strip()
    ds['label'] = norm_label(ds['label'])
    ds = ds.dropna(subset=['label', 'text'])
    ds['label'] = ds['label'].astype(int)
    # Cap at 72k
    if len(ds) > 72_000:
        ds = ds.sample(72_000, random_state=SEED)
    add_frame('WELFake', ds)
except Exception as e:
    print(f'   ERROR: {e}')

# ── 3. ErfanMoosaviMonazzah (20k) — confirmed working ──
print('3.  ErfanMoosaviMonazzah/fake-news-detection-dataset-English...')
try:
    ds = load_dataset(
        'ErfanMoosaviMonazzah/fake-news-detection-dataset-English',
        split='train'
    ).to_pandas()
    text_c  = next((c for c in ['text','content','article','news'] if c in ds.columns), None)
    label_c = next((c for c in ['label','class','fake','is_fake']  if c in ds.columns), None)
    if not text_c or not label_c:
        raise ValueError(f'Cols not found: {list(ds.columns)}')
    ds['text']  = ds[text_c].astype(str)
    ds['label'] = norm_label(ds[label_c])
    ds = ds.dropna(subset=['label', 'text'])
    ds['label'] = ds['label'].astype(int)
    if len(ds) > 20_000:
        ds = ds.sample(20_000, random_state=SEED)
    add_frame('erfan', ds)
except Exception as e:
    print(f'   ERROR: {e}')

# ── 4. redasers/difraud config=fake_news (jsonl, no custom script) ──
print('4.  redasers/difraud (fake_news config)...')
try:
    ds = load_dataset('redasers/difraud', 'fake_news', split='train').to_pandas()
    print(f'   difraud cols: {list(ds.columns)}')
    text_c  = next((c for c in ['text','content','article','body','claim'] if c in ds.columns), None)
    label_c = next((c for c in ['label','fake','is_fake','class','deceptive'] if c in ds.columns), None)
    if not text_c or not label_c:
        raise ValueError(f'Cols: {list(ds.columns)}')
    ds['text']  = ds[text_c].astype(str)
    ds['label'] = norm_label(ds[label_c])
    ds = ds.dropna(subset=['label', 'text'])
    ds['label'] = ds['label'].astype(int)
    add_frame('difraud', ds)
except Exception as e:
    print(f'   ERROR: difraud fake_news config: {e}')
    # Fallback: try mohammadjavadpirhadi (same ISOT-based data, confirmed on HF)
    print('   Trying fallback: mohammadjavadpirhadi/fake-news-detection-dataset-english...')
    try:
        ds = load_dataset(
            'mohammadjavadpirhadi/fake-news-detection-dataset-english',
            split='train'
        ).to_pandas()
        text_c  = next((c for c in ['text','content','article','news'] if c in ds.columns), None)
        label_c = next((c for c in ['label','class','fake'] if c in ds.columns), None)
        if not text_c or not label_c:
            raise ValueError(f'Cols: {list(ds.columns)}')
        ds['text']  = ds[text_c].astype(str)
        ds['label'] = norm_label(ds[label_c])
        ds = ds.dropna(subset=['label', 'text'])
        ds['label'] = ds['label'].astype(int)
        add_frame('mohammadjavad', ds)
    except Exception as e2:
        print(f'   ERROR: fallback also failed: {e2}')
        print('   WARNING:  Skipping slot 4 — enough data from other sources')

# ── 5. FEVER (80k) — confirmed working with trust_remote_code ──
print('5.  fever v1.0 (trust_remote_code=True)...')
try:
    ds = load_dataset('fever', 'v1.0', split='train',
                      trust_remote_code=True).to_pandas()
    ds['text']  = ds['claim'].astype(str)
    ds['label'] = ds['label'].map({'SUPPORTS': 0, 'REFUTES': 1, 'NOT ENOUGH INFO': -1})
    ds = ds[ds['label'] != -1].copy()
    ds['label'] = ds['label'].astype(int)
    if len(ds) > 80_000:
        ds = ds.sample(80_000, random_state=SEED)
    add_frame('FEVER', ds)
except Exception as e:
    print(f'   ERROR: {e}')

# ── 6. LIAR (12k) — trust_remote_code needed ──
print('6.  ucsbnlp/liar (trust_remote_code=True)...')
try:
    ds = load_dataset('ucsbnlp/liar', split='train',
                      trust_remote_code=True).to_pandas()
    ds['text']  = ds['statement'].astype(str)
    lmap = {'pants-fire':1,'false':1,'barely-true':1,
            'half-true':0,'mostly-true':0,'true':0}
    ds['label'] = ds['label'].map(lmap)
    ds = ds.dropna(subset=['label']).copy()
    ds['label'] = ds['label'].astype(int)
    add_frame('LIAR', ds)
except Exception as e:
    print(f'   ERROR: {e}')

print(f'\n Sources loaded: {len(frames)}')
print(f'   Total samples:  {sum(len(f) for f in frames):,}')
```

```python
# ── CELL 4: Merge + basic clean ──────────────────────────────
assert frames, 'No data loaded! Check internet is ON and rerun Cell 3.'

df = pd.concat(frames, ignore_index=True)
print(f'Raw combined: {len(df):,}')

# Basic quality filters
df = df.dropna(subset=['text', 'label'])
df['text']  = df['text'].astype(str)
df['label'] = df['label'].astype(int)
df = df[df['text'].str.len() >= 30]
df['text'] = df['text'].str.replace(r'<[^>]+>', ' ', regex=True)   # strip HTML
df['text'] = df['text'].str.replace(r'\s+', ' ',   regex=True).str.strip()
df['text'] = df['text'].str[:5000]
df = df[df['text'].str.contains(r'[a-zA-Z]{3,}', regex=True, na=False)]
df = df.drop_duplicates(subset=['text']).reset_index(drop=True)

print(f'After clean+dedup: {len(df):,}')
print(f'Fake: {df["label"].sum():,}  Real: {(df["label"]==0).sum():,}')
print(f'\nBy source:')
for src, cnt in df['source'].value_counts().items():
    print(f'  {src}: {cnt:,}')
```

```python
# ── CELL 5: Cross-source label conflict removal ───────────────
import hashlib

df['_h'] = df['text'].apply(
    lambda t: hashlib.sha256(t[:200].lower().strip().encode()).hexdigest()[:16]
)
conflict_hashes = set(
    df.groupby('_h')['label'].nunique()
    .pipe(lambda s: s[s > 1].index)
)
n_conflicts = df['_h'].isin(conflict_hashes).sum()
print(f'Label conflicts: {len(conflict_hashes):,} texts ({n_conflicts:,} rows) → removed')
df = df[~df['_h'].isin(conflict_hashes)].drop(columns=['_h']).reset_index(drop=True)
print(f'After conflict removal: {len(df):,}')
```

```python
# ── CELL 6: MinHash near-duplicate removal ───────────────────
print('Building MinHash signatures (threshold=0.85)...')
print('~3-6 min depending on dataset size...')

NUM_PERM = 64

def get_shingles(text, k=3):
    t = text.lower()
    return set(t[i:i+k] for i in range(len(t) - k + 1))

def build_minhash(text):
    m = MinHash(num_perm=NUM_PERM)
    for s in get_shingles(text[:500]):
        m.update(s.encode('utf-8'))
    return m

lsh       = MinHashLSH(threshold=0.85, num_perm=NUM_PERM)
keep_mask = np.ones(len(df), dtype=bool)

for idx, row in df.iterrows():
    key = str(idx)
    mh  = build_minhash(row['text'])
    try:
        if lsh.query(mh):
            keep_mask[idx] = False
            continue
        lsh.insert(key, mh)
    except ValueError:
        pass
    if idx % 30_000 == 0 and idx > 0:
        print(f'  {idx:,} processed, {(~keep_mask[:idx+1]).sum():,} near-dups removed')

removed = (~keep_mask).sum()
df = df[keep_mask].reset_index(drop=True)
print(f'\n After near-dedup: {len(df):,}  (removed {removed:,})')
```

```python
# ── CELL 7: Confidence-based noise filter ────────────────────
print('5-fold TF-IDF noise detection...')

vec_n = TfidfVectorizer(max_features=30_000, ngram_range=(1, 2),
                         sublinear_tf=True, min_df=3)
X     = vec_n.fit_transform(df['text'])
y     = df['label'].values
probs = np.zeros(len(df))

for fold, (tr, va) in enumerate(
    StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED).split(X, y)
):
    lr = LogisticRegression(max_iter=300, C=1.0, solver='lbfgs', n_jobs=-1)
    lr.fit(X[tr], y[tr])
    probs[va] = lr.predict_proba(X[va])[:, 1]
    print(f'  Fold {fold+1}/5 done')

noise = (
    ((df['label'] == 1) & (probs < 0.15)) |
    ((df['label'] == 0) & (probs > 0.85))
)
print(f'\nFlagged noisy: {noise.sum():,} ({noise.mean()*100:.1f}%)')
df = df[~noise].reset_index(drop=True)
print(f'After noise filter: {len(df):,}')
```

```python
# ── CELL 8: Balance + final stats ────────────────────────────
fake_n = df['label'].sum()
real_n = (df['label'] == 0).sum()
ratio  = min(fake_n, real_n) / max(fake_n, real_n)
print(f'Fake: {fake_n:,}  Real: {real_n:,}  ratio: {ratio:.3f}')

if ratio < 0.65:
    print('  Balancing (undersampling majority)...')
    n_min = min(fake_n, real_n)
    df = pd.concat([
        df[df['label'] == 0].sample(n=n_min, random_state=SEED),
        df[df['label'] == 1].sample(n=n_min, random_state=SEED),
    ]).sample(frac=1, random_state=SEED).reset_index(drop=True)
    print(f'After balance: {len(df):,} (50/50)')

print(f'\n Final dataset: {len(df):,}')
print(f'   Fake: {df["label"].sum():,}  Real: {(df["label"]==0).sum():,}')
print(f'\nBy source:')
for src, cnt in df['source'].value_counts().items():
    print(f'  {src}: {cnt:,}')
```

```python
# ── CELL 9: Split ────────────────────────────────────────────
train_df, temp_df = train_test_split(
    df, test_size=0.2, random_state=SEED, stratify=df['label'])
val_df, test_df = train_test_split(
    temp_df, test_size=0.5, random_state=SEED, stratify=temp_df['label'])

EPOCHS          = 3 if len(train_df) < 150_000 else 2
steps_per_epoch = len(train_df) // BATCH_SIZE
total_steps     = steps_per_epoch * EPOCHS
warmup_steps    = int(total_steps * 0.06)

print(f'Train: {len(train_df):,}  Val: {len(val_df):,}  Test: {len(test_df):,}')
print(f'Epochs: {EPOCHS}  Steps: ~{total_steps:,}  Warmup: {warmup_steps}')
print(f'Est.  Est: ~{total_steps*0.35/60:.0f} min on T4')
```

```python
# ── CELL 10: Tokenise ────────────────────────────────────────
print(f'Loading tokenizer: {MODEL_NAME}')
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize(examples):
    return tokenizer(
        examples['text'], truncation=True, max_length=512, padding=False
    )

def make_hf_ds(frame):
    return Dataset.from_pandas(
        frame[['text', 'label']].reset_index(drop=True)
    ).map(tokenize, batched=True, remove_columns=['text'], num_proc=2)

print('Tokenising...')
train_ds = make_hf_ds(train_df)
val_ds   = make_hf_ds(val_df)
test_ds  = make_hf_ds(test_df)
print(' Tokenisation done')
```

```python
# ── CELL 11: Label smoothing Trainer + LLRD optimiser ────────
class LabelSmoothingTrainer(Trainer):
    def __init__(self, *args, label_smoothing=0.1, **kwargs):
        super().__init__(*args, **kwargs)
        self.label_smoothing = label_smoothing

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels  = inputs.pop('labels')
        outputs = model(**inputs)
        loss    = nn.CrossEntropyLoss(label_smoothing=self.label_smoothing)(
                      outputs.logits, labels)
        return (loss, outputs) if return_outputs else loss

def get_llrd_optimizer(model, base_lr=2e-5, decay=0.9, wd=0.01):
    from torch.optim import AdamW
    no_decay = ['bias', 'LayerNorm.weight']
    groups   = []
    groups.append({
        'params': [p for n, p in model.named_parameters()
                   if 'classifier' in n or 'pooler' in n],
        'lr': base_lr, 'weight_decay': wd
    })
    n_layers = model.config.num_hidden_layers
    for li in range(n_layers - 1, -1, -1):
        lr_i   = base_lr * (decay ** (n_layers - 1 - li))
        prefix = f'roberta.encoder.layer.{li}.'
        pd_p   = [p for n, p in model.named_parameters()
                  if n.startswith(prefix) and not any(nd in n for nd in no_decay)]
        nd_p   = [p for n, p in model.named_parameters()
                  if n.startswith(prefix) and any(nd in n for nd in no_decay)]
        if pd_p: groups.append({'params': pd_p, 'lr': lr_i, 'weight_decay': wd})
        if nd_p: groups.append({'params': nd_p, 'lr': lr_i, 'weight_decay': 0.0})
    emb_lr = base_lr * (decay ** n_layers)
    groups.append({
        'params': [p for n, p in model.named_parameters() if 'embeddings' in n],
        'lr': emb_lr, 'weight_decay': wd
    })
    opt = AdamW(groups, eps=1e-8)
    print(f' LLRD: {len(groups)} groups  top={base_lr:.1e}  bottom={emb_lr:.1e}')
    return opt

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    return {
        'accuracy':  accuracy_score(labels, preds),
        'f1_macro':  f1_score(labels, preds, average='macro'),
        'precision': precision_score(labels, preds, average='macro', zero_division=0),
        'recall':    recall_score(labels, preds, average='macro', zero_division=0),
    }

print(f'Loading model: {MODEL_NAME}')
model     = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME, num_labels=2,
    id2label={0: 'REAL', 1: 'FAKE'},
    label2id={'REAL': 0, 'FAKE': 1},
)
optimizer = get_llrd_optimizer(model)
scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_steps,
)

eval_steps = min(500, steps_per_epoch)

training_args = TrainingArguments(
    output_dir=CKPT_DIR,
    num_train_epochs=EPOCHS,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE * 2,
    learning_rate=2e-5,
    warmup_steps=warmup_steps,
    weight_decay=0.01,
    fp16=True,
    max_grad_norm=1.0,
    logging_steps=100,
    eval_strategy='steps',  eval_steps=eval_steps,
    save_strategy='steps',  save_steps=eval_steps,
    load_best_model_at_end=True,
    metric_for_best_model='f1_macro',
    greater_is_better=True,
    save_total_limit=2,
    report_to='none',
    dataloader_num_workers=2,
    lr_scheduler_type='cosine',
)

trainer = LabelSmoothingTrainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    processing_class=tokenizer,
    data_collator=DataCollatorWithPadding(tokenizer=tokenizer),
    compute_metrics=compute_metrics,
    optimizers=(optimizer, scheduler),
    callbacks=[EarlyStoppingCallback(early_stopping_patience=3)],
    label_smoothing=0.1,
)
print('\n Trainer ready')
```

```python
# ── CELL 12: TRAIN ───────────────────────────────────────────
print('=' * 65)
print(' TRAINING model_B')
print('=' * 65)
print(f'  Train: {len(train_df):,}  Val: {len(val_df):,}')
print(f'  Epochs: {EPOCHS}  Batch: {BATCH_SIZE}')
print(f'  LLRD decay=0.9 | label_smooth=0.1 | grad_clip=1.0 | cosine LR')
print()

# Resume from checkpoint if session was interrupted
existing = sorted(
    [d for d in os.listdir(CKPT_DIR) if d.startswith('checkpoint-')],
    key=lambda x: int(x.split('-')[-1])
)
resume = os.path.join(CKPT_DIR, existing[-1]) if existing else None
if resume:
    print(f'Resuming... Resuming from: {resume}')

t0 = time.time()
trainer.train(resume_from_checkpoint=resume)
elapsed = time.time() - t0
print(f'\n Training complete in {elapsed/60:.1f} min')
```

```python
# ── CELL 13: Evaluate + sanity check ─────────────────────────
results     = trainer.evaluate(test_ds)
preds_out   = trainer.predict(test_ds)
pred_labels = np.argmax(preds_out.predictions, axis=-1)
true_labels = preds_out.label_ids

print('\n' + '=' * 65)
print(' model_B FINAL RESULTS')
print('=' * 65)
print(f'  Accuracy:  {results["eval_accuracy"]:.4f}  ({results["eval_accuracy"]*100:.2f}%)')
print(f'  F1 Macro:  {results["eval_f1_macro"]:.4f}')
print(f'  Precision: {results["eval_precision"]:.4f}')
print(f'  Recall:    {results["eval_recall"]:.4f}')
print()
print(classification_report(
    true_labels, pred_labels, target_names=['Real', 'Fake'], digits=4
))

# Sanity check — V1 failed these
clf = hf_pipeline(
    'text-classification',
    model=trainer.model,
    tokenizer=tokenizer,
    device=0 if torch.cuda.is_available() else -1,
)
SANITY = [
    ('COVID vaccines are safe and effective according to WHO',            'REAL'),
    ('Breaking: Scientists confirm earth is flat, NASA admits cover-up',  'FAKE'),
    ('Stock market reaches new high amid economic recovery',              'REAL'),
    ('5G towers spread coronavirus through radio waves',                  'FAKE'),
    ('Bill Gates microchip inside vaccine confirmed by documents',        'FAKE'),
    ('NASA releases new photos of a Jupiter moon discovery',              'REAL'),
    ('Scientists find water on Mars in underground lake',                 'REAL'),
    ('Government admits chemtrails are real mind control program',        'FAKE'),
]
print('\n Sanity check (need 6+/8):')
passed = 0
for text, expected in SANITY:
    r  = clf(text)[0]
    ok = r['label'] == expected
    passed += ok
    print(f'  {"" if ok else "ERROR:"} [{r["label"]} {r["score"]:.0%}] expected {expected}')
    print(f'     "{text[:68]}"')
print(f'\nSanity: {passed}/{len(SANITY)}')
```
